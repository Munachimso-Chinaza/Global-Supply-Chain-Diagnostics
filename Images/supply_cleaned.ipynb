### RFM_Values, Score

```python
supply["Total Price"] = supply["Order Item Quantity"] * supply["Order Item Total"]
```

```python
supply["order date (DateOrders)"] = pd.to_datetime(supply["order date (DateOrders)"])
supply["order date (DateOrders)"]
```

```python
late_orders = supply["order date (DateOrders)"].max()
late_orders
```

```python
customer_segment = supply.groupby("Order Customer Id").agg({
    "order date (DateOrders)" : lambda x: (late_orders - x.max()).days,
    "Order Id" : "count",
    "Total Price" : "sum"
})
customer_segment.rename(columns={"order date (DateOrders)" : "Regency",
                   "Order Id" : "Frequency",
                   "Total Price" : "Monetary"
                   }, inplace=True)

customer_segment.head()
```

```python
quantiles = customer_segment.quantile(q=[0.25,0.5,0.75])
quantiles = quantiles.to_dict()
quantiles
```

```python
def R_Score(a,b,c):
    if a <= c[b][0.25]:
        return 4
    elif a <= c[b][0.50]:
        return 3
    elif a <= c[b][0.75]:
        return 2
    else:
        return 1


def FM_Score(x,y,z):
    if x <= z[y][0.25]:
        return 1
    elif x <= z[y][0.50]:
        return 2
    elif x <= z[y][0.75]:
        return 3
    else:
        return 4
```

```python
customer_segment['R_Score'] = customer_segment['Regency'].apply(R_Score, args=('Regency',quantiles))
customer_segment['F_Score'] = customer_segment['Frequency'].apply(FM_Score, args=('Frequency',quantiles))
customer_segment['M_Score'] = customer_segment['Monetary'].apply(FM_Score, args=('Monetary',quantiles))
customer_segment.head()
```

```python
customer_segment['RFM_Score'] = customer_segment.R_Score.map(str) + customer_segment.F_Score.map(str) + customer_segment.M_Score.map(str)
customer_segment.head()
```

```python
customer_segment["RFM_Total_Score"] = customer_segment[["R_Score", "F_Score", "M_Score"]].sum(axis=1)
customer_segment["RFM_Total_Score"].unique()
```

```python
def RFM_Level(df):
    if df["RFM_Total_Score"] >= 11:
        return "Very Frequent High Value Customers"
    elif df["RFM_Total_Score"] == 10:
        return "frequently High Value Customers"
    elif df["RFM_Total_Score"] == 9:
        return "Recent Customers"
    elif df["RFM_Total_Score"] == 8:
        return "Promising"
    elif df["RFM_Total_Score"] == 7:
        return "Customers Needing Attention"
    elif df["RFM_Total_Score"] == 6:
        return "cant lose them"
    elif df["RFM_Total_Score"] ==5:
        return "At Risk"
    else:
        return "Lost Customers"

customer_segment["Customer_Segmentation"] = customer_segment.apply(RFM_Level, axis=1)
customer_segment.head()
```

```python
supply[["R_Score", "F_Score", "M_Score", "RFM_Total_Score", "Customer_Segmentation_RFM"]] = customer_segment[["R_Score", "F_Score", "M_Score", "RFM_Total_Score", "Customer_Segmentation"]]
```
